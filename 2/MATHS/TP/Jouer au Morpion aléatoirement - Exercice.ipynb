### Pour encoder le jeu, on associe à chaque joueur la liste des positions choisies. On peut ensuite représenter la grille de jeu.

```python
### par exemple: 
liste_1 = [(0,0)] # le joueur 1 a joué en position 0,0
liste_2 = [(1,1)] # le joueur 2 a joué en position 1,1
### on définit une fonction qui permet de représenter la matrice de jeu.
def encode_state(liste_1,liste_2):
    result = np.zeros((3,3))
    for z in liste_1:
        result[z[0],z[1]]=1
    for z in liste_2:
        result[z[0],z[1]]=-1
    return result
print(encode_state(liste_1,liste_2))
print("Les 1 repéresentent les positions choisies par le joueur 1, les -1 celles du joueur 2, les 0 les positions qui n'ont pas été utilisées.")
```

### On définit une fonction pour évaluer si étant donnée une liste de positions (par exemple la liste de positions choisies par le joueur 1) le joueur a gagné.

```python
def select(i):
    def f(x):
        return x[i]
    return f

def diag(x):
    return 1*(x[0]==x[1])

def antidiag(x):
    return 1*(x[1]==2-x[0])

def evaluation_victory(liste):
    result = None
    for i in range(2):
        if list(map(select(i),liste)).count(0)==3:
            return True
        if list(map(select(i),liste)).count(1)==3:
            return True
        if list(map(select(i),liste)).count(2)==3:
            return True
        if sum(list(map(diag,liste)))==3:
            return True
        if sum(list(map(antidiag,liste)))==3:
            return True
    return result
liste_1 =[(0,0),(0,1),(0,2)] ## liste de positions gagnantes
liste_2 = [(1,0),(0,1),(0,2)] ## liste de positions perdantes
print("le joueur avec la liste 1 a-t-il gagné? ",evaluation_victory(liste_1))
print("le joueur avec la liste 2 a-t-il gagné? ",evaluation_victory(liste_2))
```
