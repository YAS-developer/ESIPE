## Partie <span style="font-family: 'Times New Roman';">II</span> - Méthode de Monte-Carlo <font color="red">(partie importante)</font>.

**<font color = 'blue'>Rappel:</font> Principe de la méthode de Monte-Carlo**

En général, le problème qu'on essaie de résoudre à l'aide de les méthodes de Monte-Carlo est celui de l'estimation d'une valeur inconnue $\mu$. Pour le faire, on adopte une démarche aléatoire où l'on peut trouver une variable aléatoire $X$ et une fonction $f$ telles que l'espérance de $f(X)$ est égale à $\mu$.
$$ \mathbb{E}[f(X)] = \mu $$
Ensuite, on génère un échantillon de $n$ variables aléatoires indépendantes qui ont la même loi que la variable $X$. On estime alors la valeur de l'espérance $\mu$ en utilisant la moyenne empirique des $f(X_i)$ que l'on note $\mu_n$.
$$ \mu_n = \frac{f(X_1)+\cdots, f(X_n)}{n} $$
On dit alors que $\mu_n$ est une approximation de la valeur de $\mu$, dans le sens où elle converge vers $\mu$
$$ \mu_n = \frac{f(X_1)+\cdots+f(X_n)}{n} \rightarrow \mathbb{E}[f(X)] = \mu $$

## Partie <span style="font-family: 'Times New Roman';">III</span> - Intervalles de confiance <font color="red">(partie importante)</font>.

Etant donné des variables aléatoires $X_1,\cdots, X_n$ indépendantes et qui ont la même loi de $X$ telles que $\mathbb{E}[X] = \mu$ et $Var(X) = \sigma^2$. On sait d'après ce qui précède que
$$ \mu_n = \frac{X_1+\cdots+X_n}{n}\rightarrow \mu $$

Il faut bien comprendre que $\mu_n$ n'est qu'une apporximation de $\mu$, ainsi afin de quantifier l'erreur commise lors cette approximation, on introduit la notion des intervalles de confiance.

Un intervalle de confiance $I_\alpha$ de niveau de confiance $\alpha \in [0,1]$ est un intervalle (asymptotique) tel que $$\mathbb{P}(\mu\in I_\alpha) = \alpha.$$ Par exemple si $\alpha = 0.95$, l'intervalle $I_\alpha$ signifie que : nous sommes confiants avec une probabilité (ou un niveau de confiance) $95\%$ que la vraie valeur de $\mu$ appartient à $I_\alpha$.

#### Intervalle de confiance dans le cas où la variance $\sigma^2$ est connue.

Dans le cas où la valeur de la variance $\sigma^2$ et connue, mais que la valeur de l'espérance est inconnue, on propose de l'estimer et de lui donner un certain intervalle de confiance. On donne l'intervalle de confiance $I_\alpha$ suivant:
$$ I_\alpha = \left[\mu_n - q_\left(\frac{\alpha+1}{2}\right)\frac{\sigma}{\sqrt{n}}, \mu_n + q_\left(\frac{\alpha+1}{2}\right)\frac{\sigma}{\sqrt{n}}\right] $$

C'est à dire qu'on a
$$\mathbb{P}\left(\mu\in\left[\mu_n - q_\left(\frac{\alpha+1}{2}\right)\frac{\sigma}{\sqrt{n}}, \mu_n + q_\left(\frac{\alpha+1}{2}\right)\frac{\sigma}{\sqrt{n}}\right]\right) = \alpha $$

Pour tout $\beta\in [0,1]$, on peut calculer la valeur de $q_\beta$ grâce au code suivant (par exemple on prend $\beta = 0.8$)
~~~py
>>> from scipy.stats import norm
>>> beta = 0.8
>>> q = norm.ppf(beta)
>>> print(q)
0.8416212335729143
~~~

#### Intervalle de confiance dans le cas où la variance $\sigma^2$ est inconnue.

En général, la valeur de la variance $\sigma^2$ est aussi inconnue, et il faut donc l'estimer. Par conséquence la forme de l'intervalle de confiance change aussi. Pour la trouver, il faut d'abord calculer l'estimateur de $\sigma^2$ que l'on notera $\sigma_n^2$ et est définit par
$$ \sigma_n^2 = \frac{1}{n-1}\sum_{i=1}^n \left(X_i - \mu_n\right)^2 $$
où
$$ \mu_n = \frac{X_1+\cdots+X_n}{n}  $$

L'intervalle de confiance au niveau $\alpha$ est donnée alors par une formule légèrement differente:
$$ I_\alpha = \left[\mu_n - t_\left(\frac{\alpha+1}{2}, n-1\right)\frac{\sigma_n}{\sqrt{n}}, \mu_n + t_\left(\frac{\alpha+1}{2}, n-1\right)\frac{\sigma_n}{\sqrt{n}}\right] $$
C'est à dire que
$$\mathbb{P}\left(\mu\in\left[\mu_n - t_\left(\frac{\alpha+1}{2}, n-1\right)\frac{\sigma_n}{\sqrt{n}}, \mu_n + t_\left(\frac{\alpha+1}{2}, n-1\right)\frac{\sigma_n}{\sqrt{n}}\right]\right) = \alpha $$

Pour tout $\beta\in [0,1]$ et tout entier $m$, on peut calculer la valeur de $t_{(\beta, m)}$ grâce au code suivant (par exemple on prend $\beta = 0.8$ et $m=20$)
~~~py
>>> from scipy.stats import t
>>> beta = 0.8
>>> m = 20
>>> q = t.ppf(beta, m)
>>> print(q)
0.8599644388738727
~~~

### <font color = 'blue'>Exercice 4:</font> Intervalle de confiance pour $\pi$

En considère un carré de côté $1$ et un quart de disque inscrit dans ce carré. Si l'on tire un point au hasard dans ce carré, la probabilité que ce point soit dans le quart de disque est :
$$
\frac{\textrm{aire du quart de disque}}{\textrm{aire du carré}}.
$$
Ce rapport vaut $\displaystyle \frac{\pi}{4}$. Une façon de calculer $\pi$ consiste donc à tirer beaucoup de points au hasard dans le carré et à calculer la proportion de points qui tombent dans le quart de disque. Cette proportion multipliée par 4 donne une approximation de $\pi$.
